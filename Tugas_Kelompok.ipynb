```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

import warnings
warnings.filterwarnings("ignore")
```

1. Buatlah simulasi deret MA (1) dengan panjang pengamatan 20 dengan
nilai parameter  
θ=0.1,θ=0.3,θ=0.7,θ=1.0,θ=1.5,θ=2.0

**Simulasi Proses MA(1) dengan θ=0.1**

```python
np.random.seed(191)
ar1 = np.array([1])
ma1 = np.array([1, -0.1])
ma_object1 = ArmaProcess(ar1,ma1)
ma1 = ma_object1.generate_sample(nsample=20)
```

```python
plt.figure()
plt.plot(ma1)
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1), lag=1, c='k')
plt.xlabel(r'$Y_{t-1}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1), lag=2, c='k')
plt.xlabel(r'$Y_{t-2}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1), lag=3, c='k')
plt.xlabel(r'$Y_{t-3}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plot_acf(ma1, zero=False)
plt.show()
```
